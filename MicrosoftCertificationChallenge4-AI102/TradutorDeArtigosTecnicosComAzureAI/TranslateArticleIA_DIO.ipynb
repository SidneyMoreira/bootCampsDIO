```python
import requests
from bs4 import BeautifulSoup

def extract_text_from_url(url):
    response = requests.get(url)

    if response.status_code == 200:
      soup = BeautifulSoup(response.text, 'html.parser')
      for script_or_style in soup(["script", "style"]):
          script_or_style.decompose()
      texto = soup.get_text(separator=' ')
      lines = (line.strip() for line in texto.splitlines())
      parts = (phrase.strip() for line in lines for phrase in line.split("  "))
      texto_limpo = '\n'.join(part for part in parts if part)
      return texto_limpo
    else:
        raise ValueError(f"Failed to fetch URL. Status code: {response.status_code}")
        return None
    text = soup.get_text()
    return text


extract_text_from_url('https://dev.to/aws-builders/rag-integration-deepseeks-new-bff-in-the-ai-world-4934') 
```

```python
from langchain_openai.chat_models.azure import AzureChatOpenAI

client = AzureChatOpenAI(
    azure_endpoint = "htps do endopoint",
    api_key = "key",
    api_version = "2024-02-15-preview",
    deployment_name = "gpt-4o-mini",
    max_retrive=0
)

def translate_article(text, lang):
  messages = [
    ("sytem", "Você atua"),
    ("user", f"Traduza p {text} para o idioma {lang} e responda em markdown.")
  ]
  response = client.invoke(messages)
  print(response.content)
  return response.content

translate_article('I know your eyes in the morning Sun', 'português')


translate_article(extract_text_from_url('https://dev.to/aws-builders/rag-integration-deepseeks-new-bff-in-the-ai-world-4934'), 'pt-br')
```

```python
url = 'https://dev.to/aws-builders/rag-integration-deepseeks-new-bff-in-the-ai-world-4934'
text = extract_text_from_url(url)
article = translate_article(text, 'pt-br')

print(article)
```
